# SuperStringWithExpansion Problem

---

## Problem Statement

**Problem:** [SuperStringWithExpansion]

### Input:
- **a)** 2 disjoint alphabets called $\Sigma$ and $\Gamma = \{\gamma_1,...,\gamma_m\}$,
- **b)** a string $s \in \Sigma^*$,
- **c)** $k$ strings $t_1,...,t_k \in (\Sigma \cup \Gamma)^*$,
- **d)** and subsets $R_1,...,R_m \subseteq \Sigma^*$, all being of finite size.

### Output:
**YES** if there is a sequence of words $r_1 \in R_1, r_2 \in R_2,...,r_m \in R_m$ such that for all $i \in \{1,...,k\}$ the so-called expansion $e(t_i)$ is a substring of $s$; the expansion $e(\gamma_j)$ of the $j$-th letter $\gamma_j \in \Gamma$, $1 \leq j \leq m$, is defined by $e(\gamma_j) := r_j$, and the expansion $e(t)$ of a whole string $t \in (\Sigma \cup \Gamma)^*$ is obtained by replacing all letters from $\Gamma$ appearing within $t$ by their expansions. Otherwise output **NO**.

---

## Formal Definitions

### Substring:
We say that $v = v_1v_2 \cdots v_{\ell_v}$ is a substring of $w = w_1w_2 \cdots w_{\ell_w}$ if there is a $j$, $1 \leq j \leq \ell_w - \ell_v + 1$, such that for all $k = 1,2,...,\ell_v$ we have $v_k = w_{j+k-1}$.

### String Replacement:
Replacing the $i$-th letter $v_i$ of the string $v = v_1v_2 \cdots v_{\ell_v}$ by the string $w = w_1w_2 \cdots w_{\ell_w}$ results in the string $v_1v_2 \cdots v_{i-1}w_1w_2 \cdots w_{\ell_w}v_{i+1}v_{i+2} \cdots v_{\ell_v}$.

## List of NP-complete Problems to Choose From

### Problem: [PartitionInto3-Sets]

**Input:** A sequence $X = (x_1,x_2,...,x_{3n})$ of $3n$ natural numbers, and a natural number $B$, such that $(B/4) < x_i < (B/2)$ for all $i \in \{1,2,...,3n\}$ and $\sum_{i=1}^{3n} x_i = nB$.

**Output:** YES if $X$ can be partitioned into $n$ disjoint sets $X_1,X_2,...,X_n$ such that for all $j \in \{1,2,...,n\}$ one has $\sum_{x \in X_j} x = B$.

---

### Problem: [1-In-3-Satisfiability]

**Input:** A set of clauses $C = \{c_1,...,c_k\}$ over $n$ boolean variables $x_1,...,x_n$, where every clause contains exactly three literals.

**Output:** YES if there is a satisfying assignment such that every clause has exactly one true literal, i.e., if there is an assignment $a: \{x_1,...,x_n\} \to \{0,1\}$ such that every clause $c_j$ is satisfied and no clause has two or three satisfied literals, and NO otherwise.

---

### Problem: [MinimumCliqueCover]

**Input:** An undirected graph $G = (V,E)$ and a natural number $k$.

**Output:** YES if there is clique cover for $G$ of size at most $k$. That is, a collection $V_1,V_2,...,V_k$ of not necessarily disjoint subsets of $V$, such that each $V_i$ induces a complete subgraph of $G$ and such that for each edge $\{u,v\} \in E$ there is some $V_i$ that contains both $u$ and $v$. NO otherwise.

---

### Problem: [Graph-3-Coloring]

**Input:** An undirected graph $G = (V,E)$.

**Output:** YES if there is a 3-coloring of $G$ and NO otherwise. A 3-coloring assigns every vertex one of 3 colors such that adjacent vertices have different colors.

---

### Problem: [Longest-Common-Subsequence]

**Input:** A sequence $w_1,w_2,...,w_n$ of strings over an alphabet $\Sigma$ and a natural number $B$.

**Output:** YES if there is a string $x$ over $\Sigma$ of length $B$ which is a subsequence of all $w_i$. The answer is NO otherwise.

**Formal definition:** We say that $x = x_1x_2 \cdots x_{\ell_x}$ is a subsequence of $w = w_1w_2 \cdots w_{\ell_w}$ if there is a strictly increasing sequence of indices $i_j$, $1 \leq j \leq \ell_x$, such that for all $j = 1,2,...,\ell_x$ we have $x_j = w_{i_j}$.

---

### Problem: [MinimumRectangleTiling]

**Input:** An $n \times n$ array $A$ of non-negative numbers, positive integers $k$ and $B$.

**Output:** YES if there is a partition of $A$ into $k$ non-overlapping rectangular sub-arrays such that the sum of the entries every sub-array is at most $B$. NO otherwise.

---

### Problem: [Minimum Graph Transformation]

**Input:** Undirected graphs $G_1 = (V_1,E_1)$, $G_2 = (V_2,E_2)$ and an integer $k > 0$.

**Output:** YES if there is a transformation of order $k$ that makes $G_1$ isomorphic to $G_2$, and NO otherwise. A transformation of order $k$ removes $k$ existing edges from $E_1$ and then adds $k$ new edges to $E_1$.

---

### Problem: [MinimumDegreeSpanningTree]

**Input:** A graph $G = (V,E)$ and an integer $k$.

**Output:** YES if there is a spanning tree $T$ in which every node has degree at most $k$; NO otherwise.

# Concrete Example of Problem Inputs

Let's work through a specific example to understand the problem better:

---

## Input Components

### Alphabets:
- $\Sigma = \{a, b, c, d, e\}$ (regular alphabet)
- $\Gamma = \{A, B, D\}$ (expansion alphabet)
- $\Sigma \cap \Gamma = \emptyset$ (no overlap)

### Target string:
- $s = "abcde"$

### Strings to expand:
- $k = 2$ strings:
  - $t_1 = "Ab"$ 
  - $t_2 = "Dc"$

### Expansion sets:
- $R_A = \{a, ab\}$ (possible expansions for symbol $A$)
- $R_B = \{b, bc\}$ (possible expansions for symbol $B$) 
- $R_D = \{d, de\}$ (possible expansions for symbol $D$)

---

## Goal

Find expansions $r_A \in R_A$, $r_B \in R_B$, $r_D \in R_D$ such that:
- $e(t_1) = e("Ab") = e(A) + "b" = r_A + "b"$ is a substring of $s = "abcde"$
- $e(t_2) = e("Dc") = e(D) + "c" = r_D + "c"$ is a substring of $s = "abcde"$

---

## Solution Analysis

### Option 1:
$r_A = "a"$, $r_D = "d"$
- $e(t_1) = "a" + "b" = "ab"$ ✓ (substring of "abcde")
- $e(t_2) = "d" + "c" = "dc"$ ✗ (not a substring of "abcde")

### Option 2:
$r_A = "ab"$, $r_D = "de"$
- $e(t_1) = "ab" + "b" = "abb"$ ✗ (not a substring of "abcde")
- $e(t_2) = "de" + "c" = "dec"$ ✗ (not a substring of "abcde")

---

## Conclusion

This particular example would be a **NO** instance since no valid expansion exists.

---

## Clear Statement of the Output Condition

**Output YES if and only if there exists a choice of expansions** $r_1 \in R_1, r_2 \in R_2, \ldots, r_m \in R_m$ (one chosen word for each $\Gamma$-letter) such that, after replacing every occurrence of each $\Gamma$-letter $\gamma_j$ by its chosen expansion $r_j$ in every input string $t_i$, **every expanded string is a contiguous substring of $s$**. If no such choices exist, output NO.

### Key Points

- **Global Consistency:** The choice $r_j$ for $\gamma_j$ is global and must be used for every occurrence of $\gamma_j$ across all $t_1, \ldots, t_k$.
- **No Change to $\Sigma$ Letters:** Letters from $\Sigma$ remain unchanged during expansion.
- **Substring Requirement:** The expanded string must appear as consecutive characters (substring) somewhere inside $s$.
- **Formal Mapping:**  
    Given chosen expansions $R = (r_1, \ldots, r_m)$, define $e_R: (\Sigma \cup \Gamma)^* \to \Sigma^*$ by:
    - $e_R(a) = a$ for $a \in \Sigma$
    - $e_R(\gamma_j) = r_j$ for $\gamma_j \in \Gamma$
    - Extended to strings by concatenation: $e_R(xy) = e_R(x) e_R(y)$
- **Decision Condition:**  
    The instance is a YES-instance iff $\exists R$ with $r_j \in R_j$ for $j=1..m$ such that $\forall i \in \{1, \ldots, k\}$, $e_R(t_i)$ is a substring of $s$.

---

### Short Examples

**Example 1 (YES):**
- $\Sigma = \{a, b\}$, $\Gamma = \{A\}$
- $s = \text{"abba"}$
- $t_1 = \text{"Ab"}$
- $R_A = \{\text{"a"}, \text{"ab"}\}$
- Choose $r_A = \text{"a"}$. Then $e_R(t_1) = \text{"ab"}$, which is a substring of "abba".  
    **Output:** YES (and the expansion is A:a)

**Example 2 (NO):**
- $\Sigma = \{a, b\}$, $\Gamma = \{A\}$
- $s = \text{"abba"}$
- $t_1 = \text{"Ab"}$
- $R_A = \{\text{"aa"}\}$
- Choose $r_A = \text{"aa"}$. Then $e_R(t_1) = \text{"aab"}$, which is not a substring of "abba".  
    **Output:** NO

**Example 3 (Consistency):**
- $\Sigma = \{a, b\}$, $\Gamma = \{A\}$
- $s = \text{"abab"}$
- $t_1 = \text{"Ab"}$, $t_2 = \text{"bA"}$
- $R_A = \{\text{"a"}, \text{"ab"}\}$
- If $r_A = \text{"a"}$, $e_R(t_1) = \text{"ab"}$ (ok), $e_R(t_2) = \text{"ba"}$ (ok) $\rightarrow$ YES.
- If $r_A = \text{"ab"}$, $e_R(t_1) = \text{"abb"}$ (not in $s$) $\rightarrow$ NO.

---

### Implementation / Program Output Format

- **If no valid choices exist (or input is malformed / not in .SWE format):** Output `NO`.
- **If there is a solution:** Output a valid assignment for each $\Gamma$-letter in the format:
    ```
    A:r_A
    B:r_B
    ...
    ```
    (One assignment per line, using the uppercase letter, a colon, and the chosen expansion.)

- **If multiple solutions exist:** Any valid one is acceptable.

---

### Edge Cases & Remarks

- If some $R_j$ is empty and $\gamma_j$ appears in any $t_i$, the instance is instantly NO.
- If $\Gamma$ is empty, the problem reduces to checking whether each $t_i$ (already over $\Sigma$) is a substring of $s$.
- If $k = 0$ (no $t_i$), the decision is vacuously YES.
- If a letter from $\Gamma$ never appears in any $t_i$, its expansion can be chosen arbitrarily from $R_j$ (but the output format may still require you to print a choice).
- Checking a given candidate tuple $(r_1, \ldots, r_m)$ is straightforward: expand each $t_i$ and check substring membership in $s$ (can be done in polynomial time in input size).

The algorithm solves SuperStringWithExpansion by backtracking over all possible selections of strings from each expansion set, building a candidate concatenation step-by-step, and checking at each step whether the current prefix can still appear as a substring in the target string $S$.

Input: Target string $S$ (length $n$), and $m$ expansion groups $E_1, E_2, \dots, E_m$, where each $E_i = \{r_{i1}, \dots, r_{ik_i}\}$ is a set of non-empty strings.
Goal: Find a sequence $r_1 \in E_1, \dots, r_m \in E_m$ such that $T = r_1 \cdots r_m$ is a substring of $S$, or determine no such sequence exists.

The algorithm uses depth-first search (backtracking):

Start with empty prefix and index 0.
For the current expansion $E_i$, try each choice $r \in E_i$.
Append $r$ to the current prefix to form $P'$.
Prune early if $P'$ cannot be a substring of $S$ (heuristic).
Recurse to the next expansion.
If all $m$ expansions are chosen and the full $T$ is a substring of $S$, return YES and the solution path.
If no path succeeds, return NO.

    if pos == m:
        if prefix is substring of S:

        if PRUNE(new_prefix, pos) == false:   // heuristic

Correctness Proof
Termination

Recursion depth ≤ $m$.
Each level has finite branching ($|E_i|$).
Total nodes in recursion tree: $\leq \prod_{i=1}^m |E_i|$, which is finite.
Thus, algorithm always terminates.

Soundness

A solution is returned only when pos == m and the full prefix is a substring of $S$.
The path records exactly the choices made.
Hence, any returned solution is valid.

Completeness

The algorithm explores every possible combination of choices (one from each $E_i$).
If a valid $T$ exists, its corresponding path will be explored and detected.
Thus, if a solution exists, it will be found.

Correctness: The algorithm returns YES iff a valid expansion exists.

Running Time Analysis
Let:

$n = |S|$
$m$ = number of expansions
$k_i = |E_i|$, let $k = \max k_i$
$c$ = maximum length of any choice string

Work per node:

String concatenation: $O(c)$
Substring check (S.find(prefix)): $O(n)$

Number of nodes:

Worst case: try all combinations → $\prod k_i \leq k^m$
Since each choice is a string over $\Sigma$, and $S$ has length $n$, typically $k_i \leq n$, $m \leq n$
So: $\prod k_i \leq n^n = 2^{n \log_2 n}$

Total time:
$$T(n) = O\left( n^n \cdot n \right) = O\left( n^{n+1} \right) = 2^{O(n \log n)} = 2^{p(n)}$$
where $p(n) = (n+1) \log_2 n$ is a polynomial in $n$.
Satisfies exponential bound $2^{p(n)}$.

````python
# How to read a file in Python (https://www.dataquest.io/blog/read-file-python/)

def read_swe_file(file_path):
    with open(file_path, 'r') as swe_file:
        k = int(swe_file.readline().rstrip('\n'))

        s = swe_file.readline().rstrip('\n')

        patterns = []
        for _ in range(k):
            temp = swe_file.readline().rstrip('\n')
            last_character_was_lowercase = False

            pattern_tokens = []
            for ch in temp:
                if ch.islower():
                    if last_character_was_lowercase:
                        pattern_tokens[-1] += ch

                    else:
                        pattern_tokens.append(ch)
                        last_character_was_lowercase = True
                else:
                    pattern_tokens.append(ch)
                    last_character_was_lowercase = False

            patterns.append(pattern_tokens)

        expansion_map = {}
        for line in swe_file.readlines():
            if ':' not in line:
                continue
            
            letter, values = line.split(':', 1)
            expansion_map[letter] = values.rstrip('\n').split(',')

    return s, patterns, expansion_map
````

````python
def build_expanded_options(patterns, expansion_map):
    expanded_options = []

    for pattern in patterns:
        options_for_pattern = []

        for token in pattern:
            # include lowercase characters as single-option lists
            if isinstance(token, str) and token.islower():
                options_for_pattern.append([token])

            elif token in expansion_map:
                options_for_pattern.append(expansion_map[token])

            else:
                # unknown token -> empty option list
                options_for_pattern.append([])

        expanded_options.append(options_for_pattern)

    return expanded_options
````

### Implementation / Program Output Format

- **If no valid choices exist (or input is malformed / not in .SWE format):** Output `NO`.
- **If there is a solution:** Output a valid assignment for each $\Gamma$-letter in the format:
    ```
    A:r_A
    B:r_B
    ...
    ```
    (One assignment per line, using the uppercase letter, a colon, and the chosen expansion.)

- **If multiple solutions exist:** Any valid one is acceptable.

---

### Edge Cases & Remarks

- If some $R_j$ is empty and $\gamma_j$ appears in any $t_i$, the instance is instantly NO.
- If $\Gamma$ is empty, the problem reduces to checking whether each $t_i$ (already over $\Sigma$) is a substring of $s$.
- If $k = 0$ (no $t_i$), the decision is vacuously YES.
- If a letter from $\Gamma$ never appears in any $t_i$, its expansion can be chosen arbitrarily from $R_j$ (but the output format may still require you to print a choice).
- Checking a given candidate tuple $(r_1, \ldots, r_m)$ is straightforward: expand each $t_i$ and check substring membership in $s$ (can be done in polynomial time in input size).

````python
def find_expansions_backtrack(options_list, target_string): # Main function to find expansion assignments
    """Try to choose one option from each options_list[i] so that the concatenation
    of chosen strings is a substring of target_string. Returns list of successful
    choice-lists (paths)."""
    matches = [] # List to store matches

    if not options_list: # nothing to choose
        return matches

    def backtrack(idx, current, path):
        if idx == len(options_list):
            if current in target_string:
                matches.append(path)
            return

        for choice in options_list[idx]:
            if not choice:
                continue

            new_str = current + choice

            if target_string.find(new_str) == -1:
                continue

            backtrack(idx + 1, new_str, path + [choice])

            if matches:
                return

    backtrack(0, "", [])

    return matches
````

````python
def log_expansion_assignments(options_list, patterns, target_string):

    log = ''
    for options, pattern in zip(options_list, patterns):
        if not options:
            return "NO"

        matches = find_expansions_backtrack(options, target_string)
        log += f"Pattern: {''.join(pattern)}\n"

        if not matches:
            return "NO"

        # Print assignments for uppercase letters only
        for match in matches:
            for token, value in zip(pattern, match):
                if isinstance(token, str) and token.isupper():
                    log += f"{token}: {value}\n"

    return log
````

````python
# Simple test using the expanded options computed earlier
print(f"Using expanded options: {expanded_options}")

# Recursive helper to combine parts and check substring membership
def combine_options(parts, target_string):
    if not parts:
        return ['']

    if len(parts) == 1:
        return [p for p in parts[0] if p in target_string]

    next_parts = combine_options(parts[1:], target_string)
    combination = []

    for current_part in parts[0]:
        for next_part in next_parts:
            if current_part + next_part in target_string:
                combination.append(current_part + next_part)

    return combination

# Example: test combinations for the first pattern (if present)
if expanded_options:
    results = combine_options(expanded_options[0], target_string)
    print(f"Final Results: {results}, Count = {len(results)}")
else:
    print("No expanded options available to test")
````

````python
def find_expansions_backtrack_verbose(options_list, target_string, stop_at_first_match=False):
    print(f"Finding expansion for string '{target_string}' with expansions: {options_list}")
    matches = []

    if not options_list:
        print(f"    No expansions left to explore, returning")
        return matches

    def backtrack(idx, current, path):
        print(f"    Backtrack(idx={idx}, path={path})")

        if idx == len(options_list):
            print(f"        Reached end with current='{current}'")
            if current in target_string:
                print(f"            Match found: '{current}'")
                matches.append(current)
            return

        for choice in options_list[idx]:
            print(f"        Exploring choice: '{choice}'")
            if not choice:
                print(f"            Skipping empty choice")
                continue

            new_str = current + choice
            print(f"        New string: '{new_str}'")

            if target_string.find(new_str) == -1:
                print(f"            '{new_str}' not found in '{target_string}', skipping")
                continue

            backtrack(idx + 1, new_str, path + [choice])

            if matches and stop_at_first_match:
                print(f"            Early Stopping at first match: {matches[0]}")
                return

    backtrack(0, "", [])
    return matches

matches = find_expansions_backtrack_verbose([['a','a','a','a'],['b','b','b','b','b'],['d', 'd'], ['z']], target_string, stop_at_first_match=False)
print("\nFinal Matches:", matches, "Count =", len(matches), "Yes/No:", "Yes" if matches else "No")
````
